```python
def topla(a, b, bitsize=8):
    elde = 0
    sayi = ""
    a = a[-bitsize:]
    b = b[-bitsize:]
    maxab = max(len(a), len(b))
    a = f"{a:0>{maxab}s}"
    b = f"{b:0>{maxab}s}"
    for ai, bi in zip(a[::-1],b[::-1]):
        ab = bin(int(ai) + int(bi) + elde)[2:]
        ab = f"{ab:0>2s}"
        basamak = ab[-1]
        elde = int(ab[0])
        sayi = basamak + sayi
    return (str(elde) + sayi)[::-1][:bitsize][::-1]

def ikilik(a, bitsize=8):
    a_ = f'{bin(abs(a))[2:]:0>{bitsize}s}'[-bitsize:]
    return a_ if a>=0 else negatif(a_, bitsize=bitsize)
    

def negatif(a,bitsize=8):
    a = a[-bitsize:]
    a = f"{a:0>{bitsize}s}"
    a_ = a.replace("0", "a").replace("1","0").replace("a","1")
    return topla(a_, "1", bitsize=bitsize)

def onluk(a, signed=True, bitsize=8):
    a = a[-bitsize:]
    a = f"{a:0>{bitsize}s}"
    if signed:
        return int(a, 2) if a[0] == "0" else -int(negatif(a, bitsize=bitsize), 2)
    else:
        return int(a, 2) 
```

```python
sayi10 = 2**63-1
bitsize = 64
a = ikilik( sayi10, bitsize=bitsize)
a_ = ikilik(-sayi10, bitsize=bitsize)

aa_ = topla(a, a_, bitsize=bitsize)

print(f"{a} + {a_} = {aa_}")
print(f"{onluk(a, bitsize=bitsize)} + {onluk(a_, bitsize=bitsize)} = {onluk(aa_, bitsize=bitsize)}")
```
